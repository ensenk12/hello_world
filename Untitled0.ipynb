```python
import matplotlib.pyplot as plt
import numpy as np
```

The plot function plots the points in the graph

the label function labels the x and y axis accordingly

```python
plt.plot([1,2,3,4,5],[5,4,5,3,2],"go")
plt.title("First Plot")
plt.xlabel("X label")
plt.ylabel("Y label")
plt.show()
```
